```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression

housing = fetch_california_housing()
housing #loads the data as arrays
```

```python
fetch_california_housing(as_frame=True).frame.describe() # just to see statistics of the data
```

```python
X = housing["data"][:, [0, 1, 2, 4, 6, 7]] #making a new matrix to only include the selected features 
y = housing["target"] # y values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state=676767) #randomly splitting the data for test/training as we cannot mix them, training data is 70%
regression_line = LinearRegression().fit(X_train, y_train) #create a Linear Regression Model using the training data
regression_line.score(X_test, y_test) #testing the Linear Regression Model (LRM) using the test data
# Output = score on "how accurate" the LRM is when tested using the selected test data

```

```python
# mathematically transform = log?? to remove the outliers?? (because as we see in the statistics, AveRooms and Population have massive outliers)
from sklearn.preprocessing import FunctionTransformer
import numpy as np
X_copy = housing["data"][:, [0, 1, 2, 4, 6, 7]]

# Initialize FunctionTransformer with np.log1p
# validate=True ensures the input is finite, feature_names_out='original' for column names
log_transformer = FunctionTransformer(np.log1p, validate=True)

X_copy[:,[2, 3]] = log_transformer.fit_transform(X_copy[:,[2,3]])
X_copy #new transformed dataset
```

```python
# If we use all of the given features from the initial data,
X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(X_copy, y, test_size = 0.3, random_state=21)
regression_line = LinearRegression().fit(X_train_m, y_train_m)
regression_line.score(X_test_m, y_test_m)
# Output: slightly higher score LOL
```
