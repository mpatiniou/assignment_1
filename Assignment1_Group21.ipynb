```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression

housing = fetch_california_housing()
fetch_california_housing(as_frame=True).frame.describe()
```

```python
X_housing_s = housing["data"][:, [0, 1, 2, 4, 6, 7]] #making a new matrix to only include the selected features 
y_housing = housing["target"] # y values
X_train_housing_s, X_test_housing_s, y_train_housing_s, y_test_housing_s = train_test_split(X_housing_s, y_housing, test_size = 0.3, random_state=42) #randomly splitting the data for test/training as we cannot mix them, training data is 70%
regression_line = LinearRegression().fit(X_train_housing_s, y_train_housing_s) #create a Linear Regression Model using the training data
regression_line.score(X_test_housing_s, y_test_housing_s) #testing the Linear Regression Model (LRM) using the test data
# Output = score on "how accurate" the LRM is when tested using the selected test data
```

```python
# mathematically transform using log for the features that have a large range of values
from sklearn.preprocessing import FunctionTransformer
import numpy as np
X_housing_transformed = housing["data"][:, [0, 1, 2, 3, 6, 7]]

log_transformer = FunctionTransformer(np.log1p, validate=True)

X_housing_transformed[:,[2, 3]] = log_transformer.fit_transform(X_housing_transformed[:,[2, 3]]) #log transform skewed data

X_train_housing_t, X_test_housing_t, y_train_housing_t, y_test_housing_t = train_test_split(X_housing_transformed, y_housing, test_size = 0.3, random_state=42)
regression_line = LinearRegression().fit(X_train_housing_t, y_train_housing_t) #create a Linear Regression Model using the training data
regression_line.score(X_test_housing_t, y_test_housing_t) #testing the Linear Regression Model (LRM) using the test data
```

```python
# If we use all of the given features from the initial data,
X_train_housing_all, X_test_housing_all, y_train_housing_all, y_test_housing_all = train_test_split(housing["data"], y_housing, test_size = 0.3, random_state=42)
regression_line = LinearRegression().fit(X_train_housing_all, y_train_housing_all)
regression_line.score(X_test_housing_all, y_test_housing_all)
```

Logistic regression using wine data

```python
from sklearn.datasets import load_wine
from sklearn.linear_model import LogisticRegression
wine = load_wine(as_frame=True).frame # this loads all the data entries
wine.describe()
```

```python
wine_data = load_wine()
wine_data
X_wine = wine_data["data"][:, [0, 1, 5, 6, 9, 10, 11, 12]]
y_wine = wine_data["target"]

X_train_wine_s, X_test_wine_s, y_train_wine_s, y_test_wine_s = train_test_split(X_wine, y_wine, test_size = 0.3, random_state=42)
log_reg = LogisticRegression(random_state=67, max_iter=5000).fit(X_train_wine_s, y_train_wine_s,)
log_reg.score(X_train_wine_s, y_train_wine_s)
```

```python
X_wine_transformed = wine_data["data"][:, [0, 1, 5, 6, 9, 10, 11, 12]]

log_transformer = FunctionTransformer(np.log1p, validate=True)

X_wine_transformed[:,[4, 7]] = log_transformer.fit_transform(X_wine_transformed[:,[4, 7]])
X_wine_transformed

X_train_wine_t, X_test_wine_t, y_train_wine_t, y_test_wine_t = train_test_split(X_wine_transformed, y_wine, test_size = 0.3, random_state=21)
log_reg_t = LogisticRegression(random_state=67, max_iter = 5000).fit(X_train_wine_t, y_train_wine_t) #create a Logistic Regression Model using the training data
log_reg_t.score(X_test_wine_t, y_test_wine_t) #testing the Linear Regression Model (LRM) using the test data
```

```python
X_wine_all = wine_data["data"]
X_train_wine_all, X_test_wine_all, y_train_wine_all, y_test_wine_all = train_test_split(X_wine_all, y_wine, test_size = 0.3, random_state=21)
log_reg_all = LogisticRegression(random_state=67, max_iter=5500).fit(X_train_wine_all, y_train_wine_all)
log_reg_all.score(X_test_wine_all, y_test_wine_all)
```
